```python
import cv2
import os
```

```python
INPUT_PATH = '/data/hqvo3/Datasets/histo_cls/2_PanCancerTCGA'
```

```python
class_names = os.listdir(INPUT_PATH)
class_names.remove('LICENSE')
```

```python
len(class_names)
```

```python
first_class_name = class_names[0]
print(first_class_name)
```

```python
cls_name_folder_path = os.path.join(INPUT_PATH, first_class_name)
```

```python
mag_subfolders = os.listdir(cls_name_folder_path)
```

```python
for mag_subfolder in mag_subfolders:
    print(f'===== {mag_subfolder} =====')
    mag_subfolder_path = os.path.join(cls_name_folder_path, mag_subfolder)
    list_of_wsis = os.listdir(mag_subfolder_path)
    
    for wsi in list_of_wsis:
        print('wsi: ', wsi)
        wsi_subfolder_path = os.path.join(mag_subfolder_path, wsi)
        list_of_imgs = os.listdir(wsi_subfolder_path)
        print('=> Number of patches: ', len(list_of_imgs))
        list_of_imgs.sort()
        print(list_of_imgs[0])
        img_path = os.path.join(wsi_subfolder_path, list_of_imgs[0])
        img_mat = cv2.imread(img_path)
        print('img resolutions: ', img_mat.shape)
        break
```

```python
n_train = 216350
n_test = 55360
n_total = n_train + n_test

print(f'-> Training percentages: {(n_train/n_total)*100}')
print(f'-> Testing percentages: {(n_test/n_total)*100}')
```

```python
cnt_cases_of_each_mag = {} # cls names, then mag, then the number of cases in that mag
all_cases_of_each_mag = {} # whether same cases in different mags. cls names, then mag, then list of cases in that mag

for cls_idx, class_name in enumerate(class_names):
  print(f'--- {cls_idx} {class_name} ---')
  cls_name_folder_path = os.path.join(INPUT_PATH, class_name)
  mag_subfolders = os.listdir(cls_name_folder_path)
  mag_subfolders.sort()

  cnt_cases_of_each_mag[class_name] = {}
  all_cases_of_each_mag[class_name] = {}

  for mag_subfolder in mag_subfolders:
      mag_subfolder_path = os.path.join(cls_name_folder_path, mag_subfolder)
      list_of_wsis = os.listdir(mag_subfolder_path)
      list_of_wsis.sort()

      cnt_cases_of_each_mag[class_name][mag_subfolder] = len(list_of_wsis)
      all_cases_of_each_mag[class_name][mag_subfolder] = list_of_wsis.copy()
      
      # for wsi in list_of_wsis:
      #     wsi_subfolder_path = os.path.join(mag_subfolder_path, wsi)
      #     list_of_imgs = os.listdir(wsi_subfolder_path)
      #     list_of_imgs.sort()
      #     # print('-> len(list_of_imgs): ', len(list_of_imgs))
```
