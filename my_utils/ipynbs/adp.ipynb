```python
import os
import numpy as np
```

```python
INPUT_PATH = "/data/hqvo3/Datasets/tissue_segmentation/ADP V1.0 Release"
```

```python
SPLIT_PATH = os.path.join(INPUT_PATH, 'splits')
```

```python
npy_files = os.listdir(SPLIT_PATH)
```

```python
dict_of_mapping_from_npy_file_to_content = {}
for npy_file in npy_files:
    print(f'Read {npy_file}')
    npy_content = np.load(os.path.join(SPLIT_PATH, npy_file))
    dict_of_mapping_from_npy_file_to_content[npy_file] = npy_content
```

```python
print(dict_of_mapping_from_npy_file_to_content['train.npy'].shape)
print(dict_of_mapping_from_npy_file_to_content['valid.npy'].shape)
print(dict_of_mapping_from_npy_file_to_content['test.npy'].shape)
```

```python
total_img_files = dict_of_mapping_from_npy_file_to_content['train.npy'].shape[0] \
                  + dict_of_mapping_from_npy_file_to_content['valid.npy'].shape[0] \
                  + dict_of_mapping_from_npy_file_to_content['test.npy'].shape[0]

print('=> Total img files: ', total_img_files)
```

```python
# check unique ids and sort
all_file_ids = np.concatenate([dict_of_mapping_from_npy_file_to_content['train.npy'],
                               dict_of_mapping_from_npy_file_to_content['valid.npy'],
                               dict_of_mapping_from_npy_file_to_content['test.npy']], axis=0)

all_file_ids_uniq = np.unique(all_file_ids)
print(all_file_ids_uniq)
print(all_file_ids_uniq.shape)
```
